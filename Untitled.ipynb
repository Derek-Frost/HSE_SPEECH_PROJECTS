```python
import yaml
from torch.utils.data import DataLoader
from src.data.datasets import LJSpeechDataset
from src.data.collate import collate_fn

cfg_path = "configs/hifigan_ljspeech.yaml"
with open(cfg_path, "r", encoding="utf-8") as f:
    cfg = yaml.safe_load(f)

# ВАЖНО для Windows/Jupyter
cfg["data"]["num_workers"] = 0

mel_cfg = cfg["audio"]
ds = LJSpeechDataset(
    root=cfg["data"]["root"],
    segment_size=cfg["data"]["segment_size"],
    sr=mel_cfg["sample_rate"],
    n_fft=mel_cfg["n_fft"],
    win_length=mel_cfg["win_length"],
    hop_length=mel_cfg["hop_length"],
    n_mels=mel_cfg["n_mels"],
    fmin=mel_cfg["fmin"],
    fmax=mel_cfg["fmax"],
)

dl = DataLoader(
    ds,
    batch_size=2,
    shuffle=True,
    num_workers=0,
    collate_fn=collate_fn,
    drop_last=True,
)

batch = next(iter(dl))
batch["audio"].shape, batch["mel"].shape
```

```python
import torch
from src.model.hifigan import Generator, MultiPeriodDiscriminator, MultiScaleDiscriminator

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

mcfg = cfg["model"]
gen = Generator(
    n_mels=mel_cfg["n_mels"],
    upsample_rates=mcfg["upsample_rates"],
    upsample_kernels=mcfg["upsample_kernels"],
    resblock_kernel_sizes=mcfg["resblock_kernel_sizes"],
    resblock_dilations=mcfg["resblock_dilations"],
    channels=mcfg["channels"],
).to(device)

mpd = MultiPeriodDiscriminator().to(device)
msd = MultiScaleDiscriminator().to(device)

mel = batch["mel"].to(device)
with torch.no_grad():
    y_hat = gen(mel)  # (B,1,T)
y_hat.shape
```

```python
from src.trainer.trainer import Trainer

opt_g = torch.optim.AdamW(gen.parameters(), lr=cfg["train"]["lr_g"], betas=tuple(cfg["train"]["betas"]))
opt_d = torch.optim.AdamW(list(mpd.parameters()) + list(msd.parameters()), lr=cfg["train"]["lr_d"], betas=tuple(cfg["train"]["betas"]))

trainer = Trainer(
    gen, mpd, msd, opt_g, opt_d,
    mel_cfg=mel_cfg,
    device=device,
    lambda_fm=cfg["train"]["lambda_fm"],
    lambda_mel=cfg["train"]["lambda_mel"],
)

# Одна "эпоха" по твоему loader (может быть долго), поэтому ограничим шаги:
logs_acc = {"loss_g": 0.0, "loss_d": 0.0}
steps = 0
for batch in dl:
    # делаем один шаг через trainer.train_epoch нельзя (он прогонит всё), поэтому быстро вызовем epoch на маленьком loader:
    break

# Самый простой путь: временно сделай маленький loader
small_dl = [next(iter(dl)) for _ in range(10)]
logs = trainer.train_epoch(small_dl)
logs
```
